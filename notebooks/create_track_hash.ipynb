```python
import json
import csv

track_translations = {}

# CSV scraping
with open("ocr_track_names.csv", newline="") as csvfile:
    reader = csv.DictReader(csvfile)
    for row in reader:
        for key in ('normal', 'gray'):
            if not row[key] in track_translations:
                track_translations[row[key]] = row['track']
    

with open("track_translations.json", "w") as f:
    json.dump(track_translations, f)
```

```python
with open("track_translations.json", "r") as f:
    data = json.load(f)
```
